```python
arlie_folder = r"T:\DLR-DFD\Analysis3\Input\ARLIE\files"
# open all files that end with arlie.csv and concatenate them into a single dataframe
arlie_dfs = []
for file in os.listdir(arlie_folder):
    if file.endswith("arlie.csv"):
        df = pd.read_csv(os.path.join(arlie_folder, file), sep = ";")
        arlie_dfs.append(df)

# concatenate all dataframes in the list into a single dataframe
arlie_df = pd.concat(arlie_dfs, ignore_index=True)
```

```python
print("Number of Lakes and rivers within Arlie:", len(arlie_df['river_km_id'].unique()))
```
